## Content‑Based “Because You Watched…” Recommender

#### Step 1 — Imports & Load Cleaned Data

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MultiLabelBinarizer, OneHotEncoder, StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.neighbors import NearestNeighbors

df = pd.read_csv("data/clean/titles_clean.csv")
```

#### Step 2 — TF‑IDF Vectorization of Description
This captures semantic meaning of each title.

```python
tfidf = TfidfVectorizer(stop_words="english", max_features=5000)
tfidf_matrix = tfidf.fit_transform(df["description"])
```

#### Step 3 — Encode Genres, Country, Rating Group
Genres → Multi‑Hot Encoding

```python
df["genre_list"] = df["listed_in"].str.split(", ")

mlb = MultiLabelBinarizer()
genre_matrix = mlb.fit_transform(df["genre_list"])
```

Primary Country → One‑Hot

```python
ohe_country = OneHotEncoder(sparse_output=False)
country_matrix = ohe_country.fit_transform(df[["primary_country"]])
```

Rating Group → One‑Hot

```python
ohe_rating = OneHotEncoder(sparse_output=False)
rating_matrix = ohe_rating.fit_transform(df[["rating_group"]].fillna("Unknown"))
```

#### Step 4 — Numeric Features

```python
numeric_features = df[[
    "runtime_min",
    "n_seasons",
    "cast_size",
    "description_length",
    "n_genres"
]].fillna(0)

scaler = StandardScaler()
numeric_matrix = scaler.fit_transform(numeric_features)
```

#### Step 5 — Combine All Features Into One Matrix
This is our full content embedding for each title.

```python
from scipy.sparse import hstack

feature_matrix = hstack([
    tfidf_matrix,
    genre_matrix,
    country_matrix,
    rating_matrix,
    numeric_matrix
])
feature_matrix = feature_matrix.tocsr()
```

#### Step 6 — Build Similarity Engine
We use Nearest Neighbors (KNN) with cosine similarity.

```python
nn = NearestNeighbors(metric="cosine", algorithm="brute")
nn.fit(feature_matrix)
```

#### Step 7 — Recommendation Function

```python
def recommend(title, n=10):
    if title not in df["title"].values:
        return f"Title '{title}' not found in dataset."

    idx = df.index[df["title"] == title][0]
    distances, indices = nn.kneighbors(feature_matrix[idx], n_neighbors=n+1)

    results = []
    for dist, i in zip(distances[0][1:], indices[0][1:]):
        results.append((df.loc[i, "title"], 1 - dist))

    return results
```

#### Let's try it now

```python
recommend("Kota Factory", n=5)
```

This system looks at what Kota Factory is about — its themes, genres, cast size, country, description, and other metadata — and finds other shows that share similar characteristics.

The number next to each show is how similar it is.

For example, Bhaag Beanie Bhaag has a similarity score of 0.859, meaning it’s very close to Kota Factory in terms of content style.
